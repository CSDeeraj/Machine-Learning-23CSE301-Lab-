## Aim
To identify the fruit most likely to be selected from a fruit basket by applying the majority-voting principle used in a Random Forest.

## Problem Statement
Multiple samples are drawn from a fruit basket. A separate decision tree is built for each sample, and each tree predicts a fruit class. The final class is selected by majority voting. Based on the predictions illustrated, determine which fruit is most likely to be taken.

## Requirements
- Python 3
- pandas
- matplotlib *(optional, for visualization)*

## Concept
A **Random Forest** is an ensemble-learning method that combines predictions from many decision trees. Each tree is trained using a different bootstrap sample of the training data. For a classification task, the trees vote for a class, and the class receiving the most votes is selected.

**Majority-voting rule:**

`Final class = class with the highest vote count`

The figure shows Apple from Tree-1 and Tree-2, and Banana from Tree-n. The notebook uses those illustrated outputs. If additional tree outputs are supplied, add them to the prediction list before determining the final class.

## Procedure
1. Import the required library.
2. Enter the fruit predictions shown in the figure.
3. Count how many trees voted for each fruit.
4. Select the fruit with the highest vote count.
5. Display the result and interpret it.

## Code

### Step 1 — Import the required library

```python
#CH.SC.U4CSE24106
import pandas as pd
```

### Step 2 — Enter the decision-tree predictions

```python
# Predictions represented in the exercise figure
predictions = ['Apple', 'Apple', 'Banana']

print('Tree predictions:', predictions)
print('Number of predictions:', len(predictions))
```

### Step 3 — Count the votes for each fruit

```python
votes = pd.Series(predictions).value_counts()
votes
```

### Step 4 — Determine the majority-vote class

```python
highest_votes = votes.max()
winners = votes[votes == highest_votes].index.tolist()

if len(winners) == 1:
    final_class = winners[0]
    print('Final predicted fruit:', final_class)
else:
    print('Tie between:', ', '.join(winners))
```

## Output / Observation
The vote count for the illustrated predictions is:

| Fruit | Votes |
|---|---:|
| Apple | 2 |
| Banana | 1 |

Apple receives more votes than Banana, so it wins the majority vote.

## Result
**Apple is the fruit most likely to be taken**, based on the decision-tree predictions shown in the figure. It receives 2 of the 3 illustrated votes.

## Optional 1 — Visualize the vote distribution
Run this cell to create a bar chart of the decision-tree votes.

```python
import matplotlib.pyplot as plt

votes.sort_values(ascending=False).plot(kind='bar', title='Random Forest — Fruit Votes')
plt.xlabel('Fruit')
plt.ylabel('Number of Votes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
